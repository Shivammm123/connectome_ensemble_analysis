# Interactive Connectome Analysis Dashboard

Explore connectivity similarity analysis interactively!

**Based on:** Liessem et al. (2025) methodology

## Setup

```python
import sys
sys.path.append('../scripts')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import yaml

from utils.data_loader import ConnectomeDataLoader
from utils.similarity import build_connectivity_vectors, cosine_similarity_matrix, find_top_similar

# For interactive widgets
try:
    from ipywidgets import interact, IntSlider, FloatSlider, Dropdown
    WIDGETS_AVAILABLE = True
except ImportError:
    print("Install ipywidgets for interactive controls: pip install ipywidgets")
    WIDGETS_AVAILABLE = False

# Set style
sns.set_style('whitegrid')
plt.rcParams['figure.dpi'] = 100

print("✓ Imports complete")
```

## 1. Load Data

```python
# Load configuration
with open('../config.yaml', 'r') as f:
    config = yaml.safe_load(f)

# Initialize data loader
loader = ConnectomeDataLoader('../config.yaml')

# Load all data
print("Loading data...")
connections, neurons, name_mapping = loader.load_all_data(verbose=True)
print("\n✓ Data loaded!")
```

```python
# Quick data overview
print(f"Total connections: {len(connections):,}")
print(f"Unique source neurons: {connections['source'].nunique():,}")
print(f"Unique target neurons: {connections['target'].nunique():,}")
print(f"Total neurons with attributes: {len(neurons):,}")
```

## 2. Explore Single Neuron Analysis Results

```python
# Load previous analysis results
results_file = Path('../results/single_neuron/similarity_results.csv')

if results_file.exists():
    results = pd.read_csv(results_file)
    print(f"✓ Loaded {len(results)} similar neurons")
    print("\nTop 10 most similar:")
    display(results.head(10))
else:
    print("⚠ No results found. Run single_neuron_analysis.py first!")
```

## 3. Interactive Similarity Explorer

```python
# Interactive plot of top N similar neurons
if WIDGETS_AVAILABLE and results_file.exists():
    
    def plot_top_similar(top_n=10, min_similarity=0.0):
        """Interactive plot of similar neurons."""
        
        # Filter and get top N
        filtered = results[results['similarity'] >= min_similarity]
        top = filtered.head(top_n)
        
        if len(top) == 0:
            print(f"No neurons with similarity >= {min_similarity}")
            return
        
        # Create plot
        fig, ax = plt.subplots(figsize=(10, 6))
        
        bars = ax.barh(range(len(top)), top['similarity'])
        
        # Color by similarity
        colors = plt.cm.viridis(top['similarity'])
        for bar, color in zip(bars, colors):
            bar.set_color(color)
        
        ax.set_yticks(range(len(top)))
        ax.set_yticklabels([f"Neuron {int(nid)}" for nid in top['neuron_id']])
        ax.set_xlabel('Cosine Similarity', fontweight='bold')
        ax.set_title(f'Top {len(top)} Similar Neurons (threshold: {min_similarity})',
                    fontweight='bold')
        ax.grid(axis='x', alpha=0.3)
        
        # Add values
        for i, val in enumerate(top['similarity']):
            ax.text(val + 0.01, i, f'{val:.3f}', va='center')
        
        plt.tight_layout()
        plt.show()
        
        print(f"\nShowing {len(top)} neurons")
        print(f"Mean similarity: {top['similarity'].mean():.3f}")
    
    # Interactive controls
    interact(plot_top_similar,
             top_n=IntSlider(min=5, max=30, step=5, value=10, description='Top N:'),
             min_similarity=FloatSlider(min=0.0, max=0.9, step=0.1, value=0.0, description='Min similarity:'))
else:
    print("Install ipywidgets for interactive controls: pip install ipywidgets")
```

## 4. Distribution Analysis

```python
if results_file.exists():
    fig, axes = plt.subplots(1, 3, figsize=(15, 4))
    
    # Histogram
    axes[0].hist(results['similarity'], bins=30, edgecolor='black', alpha=0.7)
    axes[0].axvline(results['similarity'].mean(), color='red', linestyle='--', 
                    label=f'Mean: {results["similarity"].mean():.3f}')
    axes[0].set_xlabel('Similarity Score')
    axes[0].set_ylabel('Frequency')
    axes[0].set_title('Distribution of Similarities')
    axes[0].legend()
    axes[0].grid(alpha=0.3)
    
    # Box plot
    axes[1].boxplot(results['similarity'])
    axes[1].set_ylabel('Similarity Score')
    axes[1].set_title('Similarity Statistics')
    axes[1].grid(alpha=0.3)
    
    # Cumulative
    sorted_sim = np.sort(results['similarity'])
    cumulative = np.arange(1, len(sorted_sim)+1) / len(sorted_sim)
    axes[2].plot(sorted_sim, cumulative, linewidth=2)
    axes[2].set_xlabel('Similarity Score')
    axes[2].set_ylabel('Cumulative Probability')
    axes[2].set_title('Cumulative Distribution')
    axes[2].grid(alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    # Statistics
    print("\nSimilarity Statistics:")
    print(f"  Mean: {results['similarity'].mean():.4f}")
    print(f"  Median: {results['similarity'].median():.4f}")
    print(f"  Std: {results['similarity'].std():.4f}")
    print(f"  Min: {results['similarity'].min():.4f}")
    print(f"  Max: {results['similarity'].max():.4f}")
    print(f"  Q25: {results['similarity'].quantile(0.25):.4f}")
    print(f"  Q75: {results['similarity'].quantile(0.75):.4f}")
```

## 5. Analyze Any Neuron On-The-Fly

```python
def quick_similarity_analysis(neuron_id, min_synapses=3, top_n=10):
    """
    Quick similarity analysis for any neuron.
    
    Parameters
    ----------
    neuron_id : int
        Neuron to analyze
    min_synapses : int
        Minimum synapse threshold
    top_n : int
        Number of similar neurons to return
    """
    
    print(f"Analyzing neuron: {neuron_id}")
    print(f"Name: {loader.get_neuron_name(neuron_id)}")
    
    # Filter connections
    filtered_conn = loader.filter_connections(min_synapses=min_synapses, verbose=False)
    
    # Get query connectivity
    query_conn = loader.get_neuron_connectivity(neuron_id, filtered_conn)
    
    if len(query_conn) == 0:
        print("❌ No connections found!")
        return None
    
    print(f"\nConnectivity:")
    print(f"  Connections: {len(query_conn)}")
    print(f"  Total synapses: {query_conn['weight'].sum()}")
    print(f"  Targets: {query_conn['target'].nunique()}")
    
    # Find comparison neurons
    query_targets = set(query_conn['target'].unique())
    shared_conn = filtered_conn[filtered_conn['target'].isin(query_targets)]
    candidates = shared_conn['source'].unique()
    candidates = [n for n in candidates if n != neuron_id]
    candidates = candidates[:200]  # Limit
    
    all_neurons = [neuron_id] + candidates
    analysis_conn = filtered_conn[filtered_conn['source'].isin(all_neurons)]
    
    # Compute similarity
    print(f"\nComputing similarity to {len(candidates)} neurons...")
    conn_matrix, _ = build_connectivity_vectors(analysis_conn, neuron_ids=all_neurons)
    sim_matrix = cosine_similarity_matrix(conn_matrix)
    
    # Find top similar
    query_idx = 0
    top_similar = find_top_similar(query_idx, sim_matrix, all_neurons, top_n=top_n)
    
    # Add names
    top_similar = loader.add_names_to_dataframe(top_similar, ['neuron_id'])
    
    print(f"\n✓ Top {len(top_similar)} similar neurons:")
    display(top_similar)
    
    # Quick plot
    fig, ax = plt.subplots(figsize=(10, 5))
    ax.barh(range(len(top_similar)), top_similar['similarity'])
    ax.set_yticks(range(len(top_similar)))
    ax.set_yticklabels([f"{row['neuron_id_name'][:20]}" if 'neuron_id_name' in top_similar.columns 
                        else f"Neuron {int(row['neuron_id'])}" 
                        for _, row in top_similar.iterrows()])
    ax.set_xlabel('Similarity')
    ax.set_title(f'Top Similar to {loader.get_neuron_name(neuron_id)}')
    plt.tight_layout()
    plt.show()
    
    return top_similar

print("✓ Function ready. Use: quick_similarity_analysis(neuron_id)")
```

```python
# Try it! Replace with any neuron ID
neuron_to_analyze = 720575941510642192  # Change this!

results_quick = quick_similarity_analysis(
    neuron_to_analyze,
    min_synapses=3,
    top_n=10
)
```

## 6. Compare Two Neurons

```python
def compare_two_neurons(neuron_a, neuron_b, min_synapses=3):
    """Compare connectivity between two neurons."""
    
    filtered_conn = loader.filter_connections(min_synapses=min_synapses, verbose=False)
    
    conn_a = loader.get_neuron_connectivity(neuron_a, filtered_conn)
    conn_b = loader.get_neuron_connectivity(neuron_b, filtered_conn)
    
    targets_a = set(conn_a['target'].unique())
    targets_b = set(conn_b['target'].unique())
    
    shared = targets_a & targets_b
    only_a = targets_a - targets_b
    only_b = targets_b - targets_a
    
    print(f"Neuron A: {loader.get_neuron_name(neuron_a)} ({neuron_a})")
    print(f"Neuron B: {loader.get_neuron_name(neuron_b)} ({neuron_b})")
    print(f"\nTarget overlap:")
    print(f"  Shared targets: {len(shared)}")
    print(f"  Only in A: {len(only_a)}")
    print(f"  Only in B: {len(only_b)}")
    print(f"  Jaccard index: {len(shared)/(len(targets_a | targets_b)):.3f}")
    
    # Venn diagram
    from matplotlib_venn import venn2
    fig, ax = plt.subplots(figsize=(8, 6))
    venn2([targets_a, targets_b], 
          set_labels=(f'Neuron A\n{loader.get_neuron_name(neuron_a)[:15]}', 
                     f'Neuron B\n{loader.get_neuron_name(neuron_b)[:15]}'))
    plt.title('Target Overlap')
    plt.show()

print("✓ Function ready. Use: compare_two_neurons(neuron_a, neuron_b)")
```

## 7. Export Custom Results

```python
# Save any analysis results
def save_custom_results(dataframe, filename):
    """Save results to custom location."""
    output_dir = Path('../results/custom')
    output_dir.mkdir(parents=True, exist_ok=True)
    
    filepath = output_dir / filename
    dataframe.to_csv(filepath, index=False)
    print(f"✓ Saved: {filepath}")

print("✓ Function ready. Use: save_custom_results(df, 'my_analysis.csv')")
```
